```python
import pandas as pd
from sklearn.model_selection import train_test_split

MIN_PRICE = 5_000
MAX_PRICE = 1_000_000

df = pd.read_csv("../data/raw/cars_dataframe.csv")
print("Départ :", df.shape)   # attendu : (101896, 15)
```

```python
n0 = len(df)

# 1a) lignes sans prix : on ne remplace jamais la cible
df = df[df["Price"].notna()]
n1 = len(df)

# 1b) prix hors des seuils
df = df[df["Price"].between(MIN_PRICE, MAX_PRICE)]
n2 = len(df)

print(f"Sans prix retirées : {n0 - n1}")        # attendu : 25 483
print(f"Prix invalides retirées : {n1 - n2}")   # attendu : 966
print("Restant :", n2)                          # attendu : 75 447
```

```python
n_dup = df.duplicated().sum()
print("Doublons exacts :", n_dup)

df = df.drop_duplicates().reset_index(drop=True)
print("Restant :", len(df))
```

```python
df.to_csv("../data/processed/cars_clean.csv", index=False)
```

```python
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

train_df.to_csv("../data/processed/train.csv", index=False)
test_df.to_csv("../data/processed/test.csv", index=False)

print("Train :", train_df.shape, "| Test :", test_df.shape)
```
