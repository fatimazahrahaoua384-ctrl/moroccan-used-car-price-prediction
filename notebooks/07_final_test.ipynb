```python
import sys; sys.path.append("..")
import numpy as np
import pandas as pd
from src.features import add_features
from src.model import build_model

train = add_features(pd.read_csv("../data/processed/train.csv"))
test = add_features(pd.read_csv("../data/processed/test.csv"))

X_train, y_train = train.drop(columns="Price"), train["Price"]
X_test, y_test = test.drop(columns="Price"), test["Price"]

model = build_model()
model.fit(X_train, y_train)      # apprentissage sur le train seulement
pred = model.predict(X_test)

abs_err = np.abs(pred - y_test)
dummy_err = np.abs(y_train.median() - y_test)

print("Taille du test :", len(test))
print("MAE test :", round(abs_err.mean()))
print("Erreur absolue médiane :", round(abs_err.median()))
print("Erreur relative médiane (%) :", round((abs_err / y_test * 100).median(), 1))
print("MAE dummy (médiane du train) :", round(dummy_err.mean()))
```
