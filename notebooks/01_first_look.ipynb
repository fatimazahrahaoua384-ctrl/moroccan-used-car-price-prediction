```python
p = df["Price"]
print("prix == 1       :", (p == 1).sum())
print("prix < 1 000    :", (p < 1_000).sum())
print("prix < 5 000    :", (p < 5_000).sum())
print("prix > 5 M      :", (p > 5e6).sum())
print("prix > 10 M     :", (p > 1e7).sum())
print("prix > 100 M    :", (p > 1e8).sum())
```

```python
# Les prix bas (hors 1) : sont-ils crédibles ?
df.loc[(p > 1) & (p < 5_000), ["Brand", "Model", "Year", "Price"]].sort_values("Price").head(20)
```

```python
# La zone haute : voitures de luxe légitimes ou erreurs ?
df.loc[p.between(1e6, 1e8), ["Brand", "Model", "Year", "Price"]].sort_values("Price", ascending=False).head(30)
```
