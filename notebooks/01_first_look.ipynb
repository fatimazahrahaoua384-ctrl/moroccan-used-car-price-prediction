```python
MIN_PRICE = 5_000
n_avec_prix = df["Price"].notna().sum()
n_sous_seuil = (df["Price"] < MIN_PRICE).sum()
print(n_sous_seuil, round(n_sous_seuil / n_avec_prix * 100, 2), "%")
```

**Décision : seuil minimum de prix = 5 000 MAD.**
Inspection de 30 lignes entre 1 000 et 5 000 MAD : surtout de vieilles voitures
(avant 2005) à prix plausibles, mais quelques erreurs (Audi RS3 2024 à 2024 MAD,
Clio 2021 à 1 600). Aucune valeur ronde dominante. 557 lignes retirées (0,7 %). 

```python
p = df["Price"]
lux = df.loc[p.between(1e6, 5e6), ["Brand", "Model", "Year", "Mileage", "Price"]]
print(len(lux))
print(lux["Brand"].value_counts().head(15))
lux.sort_values("Price").head(20)
```

```python
# 1) Haut de la zone : le luxe monte-t-il jusqu'à 5 M ou s'arrête-t-il avant ?
lux.sort_values("Price", ascending=False).head(20)

# 2) Marques économiques : où commencent les prix anormaux ?
eco = ["Dacia", "Renault", "Fiat", "Peugeot", "Citroen"]
d = df[df["Brand"].isin(eco)]
print(d["Price"].describe(percentiles=[.5, .9, .95, .99]))
print(d["Price"].between(300_000, 1_000_000).sum(), (d["Price"] > 1_000_000).sum())

# 3) Au-delà de 5 M : vérifier que ce sont bien des chiffres bidon
df.loc[p > 5e6, ["Brand", "Model", "Year", "Price"]].sample(15, random_state=42)
```

```python
MIN_PRICE = 5_000
MAX_PRICE = 1_000_000

n_avec_prix = df["Price"].notna().sum()
hors = ~df["Price"].between(MIN_PRICE, MAX_PRICE) & df["Price"].notna()
print(hors.sum(), round(hors.sum() / n_avec_prix * 100, 2), "%")
```

**Décision : seuil maximum de prix = 1 000 000 MAD.**
Au-dessus de 5 M : 100 % d'erreurs (échantillon de 15). Entre 1 et 5 M : moitié luxe
plausible, mais beaucoup d'erreurs sur les marques économiques (Logan, Clio, Dokker,
Fiat 500 à plus d'1 M). Les voitures à plus d'1 M MAD sont hors périmètre du modèle.
